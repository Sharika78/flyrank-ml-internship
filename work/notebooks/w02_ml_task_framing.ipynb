## 5. Why ML Beats a Fixed Rule Here

- **Limitations of Fixed Rules:** A naive hand-written rule (e.g., *"Flag any page if clicks drop by more than 20% month-over-month"*) applies a rigid, blunt threshold universally. It fails to account for seasonal variations, historical page volatility, and multi-variable interactions.
- **Advantages of Machine Learning:**
  1. **Multi-Signal Synthesis:** ML can weigh dozens of features simultaneously (impressions, click-through rates, historical ranking volatility, content age) rather than relying on a single metric.
  2. **Reduced False Positives:** ML learns non-linear patterns to distinguish between a natural seasonal traffic dip and actual structural content decay.
  3. **Measurable Lift:** Empirical tests on the FlyRank starter dataset show that a learned model achieves roughly a $3\times$ lift in Precision@50 compared to a static baseline rule (~0.24 to ~0.74).
